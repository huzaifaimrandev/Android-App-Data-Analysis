```python
from re import L
def clean_label(label):
    label = label.lower()
    label = label.replace(' ','_')
    label = label.replace('current_version','current_version')
    label = label.replace('android_version','android_version')
    return label


```

```python
dataset['category'] = dataset['category'].str.title()
```

```python
dataset['category'] = dataset['category'].str.replace('And','&')
```

```python
dataset['android_versionsion'] = dataset['android_versionsion'].str.replace('Varies with device','Unknow')
dataset['size'] = dataset['size'].str.replace('Varies with device','Unknown')
dataset['installs'] = dataset['installs'].str.replace('+','').str.replace(',','')
dataset['price'] = dataset['price'].str.replace('$','')
dataset['rating'] = dataset['rating'].astype(float)
dataset['price'] = dataset['price'].astype(float)
dataset.rename({'price':'price_usd'},axis=1, inplace = True)
```

```python
def category(element):
  if element > 100000:
    return 'Most high reviewed app'
  elif element < 100000 and element >= 20000:
    return 'High reviewed app'
  elif element < 20000 and element >= 5000:
    return 'Average reviewed app'
  else:
      return 'Low reviewed app'
```

```python
dataset['reviews'].map(category).value_counts()
```

```python
dataset['installs'] = dataset['installs'].astype(int)
```

```python
def category(element):
  if element > 100000 :
    return 'Most downloaded app'
  elif element < 100000 and element >= 40000:
    return 'High downloaded app'
  elif element < 40000 and element >= 7000:
    return 'Average downloaded app'
  else:
     return 'Low downloaded app'
```
